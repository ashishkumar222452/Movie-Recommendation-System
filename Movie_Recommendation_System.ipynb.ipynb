```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import zipfile

with zipfile.ZipFile("/content/ml-latest-small.zip", 'r') as zip_ref:
    zip_ref.extractall("data")
```

```python
movies = pd.read_csv("/content/data/ml-latest-small/movies.csv")
movies.head()
```

```python
vectorizer = CountVectorizer(tokenizer=lambda x: x.split('|'))
genre_matrix = vectorizer.fit_transform(movies['genres'])
```

```python
cosine_sim = cosine_similarity(genre_matrix, genre_matrix)
```

```python
def recommend_movie(movie_title, n=5):
    if movie_title not in movies['title'].values:
        print("Movie not found in dataset")
        return

    index = movies[movies['title'] == movie_title].index[0]
    similarity_scores = list(enumerate(cosine_sim[index]))
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)

    print(f"Movies similar to '{movie_title}':")
    for i in similarity_scores[1:n+1]:
        print(movies.iloc[i[0]].title)
```

```python
recommend_movie("Toy Story (1995)")
```
